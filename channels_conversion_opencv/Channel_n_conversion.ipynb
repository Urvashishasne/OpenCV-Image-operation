# openCV : image channels 

```python
import cv2 
import matplotlib.pyplot as plt
```

Matplotlib expects a image in RGB where OpenCV stores an image in BGR format. 

```python
img = cv2.imread("images\elephant.png") 
#cv2.imshow("original image",img)
print(img.shape) 
plt.imshow(img) 
```

```python
#for correct display we need to reverse the channels of image
revers_img=img[:,:,::-1]
print("Original Image :")
plt.imshow(revers_img) 
```

1. BGR TO RGB

```python
img_RGB = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
print("RGB color channel :")
plt.imshow(img_RGB)
```

2. RGB TO BGR 

```python
img_BGR = cv2.cvtColor(img_RGB,cv2.COLOR_RGB2BGR)
print("BGR color channel :")
plt.imshow(img_BGR)
#cv2.imshow("ori.imge",img1)
```

3. HSV (Hue Saturation  Value ) color space : BGR to HSV

```python
img_HSV=cv2.cvtColor(img_BGR,cv2.COLOR_BGR2HSV)
print("HSV color space")
plt.imshow(img_HSV)
```

Hue represent the color of image    
saturation represent the intensity of color     
value represent the value   

```python
H,S,V= cv2.split(img_HSV) # split channels
plt.figure(figsize=[20,5])
#plt.imshow(img_BGR);plt.title("BGR img")
plt.imshow(H);plt.title("H Channel")
plt.imshow(S);plt.title("S Channel")
plt.imshow(V);plt.title("V Channel")
```

modifying individual channel 

```python
h_new = H+10
img_merged= cv2.merge((h_new,S,V))
plt.imshow(img_merged)
```

```python
# convert merged image into rgb 
img_rgb_merged=cv2.cvtColor(img_merged,cv2.COLOR_HSV2RGB)
plt.imshow(img_rgb_merged)
```

```python
# save the image
cv2.imwrite("edited image.png",img_rgb_merged)
```
